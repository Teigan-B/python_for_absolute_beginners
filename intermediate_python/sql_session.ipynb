```python
with engine_903.connect() as con:
    stmt = "SELECT * FROM episodes"
    result_proxy = con.execute(text(stmt))
    results = result_proxy.fetchall()

#print all:
results

#first row - can use index to get stuff out, making slices from lists/dictionaries/dataframes. Index starts from 0 so first row is at index locaton 0
first_row = results[0]
print(first_row)
print(first_row.CHILD)
```

create header table for 903
if wanted to create table for children with no ethnicity code REFU and NOBT
make a select statement, use the WHERE method from sqlalchemy (did not need to import at start because included as standard; is part of select)

double equals because not variable
can use OR function for REFU or NOBT
or function is or_()
reason it has underscore is because or keyword already exists in python as a keyword (same as 'and'). So creators of sqlalchemy could not just have 'or'; instead used 'or_' 
if wanted to use and, and_


```python
header = Table("header", metadata_903, autoload_with=engine_903)

with engine_903.connect() as con:
    stmt = select(header).where(or_(header.columns.ETHNIC== "REFU",
                                header.columns.ETHNIC== "NOBT"))
    #print(stmt)
    result_proxy = con.execute(stmt)
    result = result_proxy.fetchmany(size = 20)

result

```

make or_ statement where artist id is 1 or 2. select all of the albums table where the ArtistId is 1 or 2

```python
albums = Table("albums", metadata_chinook, autoload_with=engine_chinook)

with engine_chinook.connect() as con:
    stmt = select(albums).where(or_(albums.columns.ArtistID== 1,
                                    albums.columns.ArtistID== 2))
    result_proxy = con.execute(stmt)
    result = result_proxy.fetchmany(size = 10)

albums_df = pd.DataFrame(result)



# with engine_chinook.connect() as con:
#     stmt = select(albums).where(or_(albums.columns.ArtistId == 1,
#                                     albums.columns.ArtistId == 2))
#     result_proxy = con.execute(stmt)
#     result = result_proxy.fetchmany(size=10)

# albums_df = pd.DataFrame(result)
# albums_df
 
```
